```python
from torch.utils.data import Dataset

class MELDDataset(Dataset):

    def __init__(self, text, audio, video, labels):

        self.text = text
        self.audio = audio
        self.video = video
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):

        return (
            self.text[idx],
            self.audio[idx],
            self.video[idx],
            self.labels[idx]
        )
```

```python
class Projection(nn.Module):

    def __init__(self, input_dim, hidden_dim=256):

        super().__init__()

        self.layer = nn.Sequential(

            nn.Linear(input_dim, hidden_dim),

            nn.ReLU(),

            nn.Dropout(0.3)

        )

    def forward(self, x):

        return self.layer(x)
```

```python
class ContextTransformer(nn.Module):

    def __init__(self, dim=256):

        super().__init__()

        encoder = nn.TransformerEncoderLayer(

            d_model=dim,

            nhead=8,

            batch_first=True

        )

        self.transformer = nn.TransformerEncoder(

            encoder,

            num_layers=2
        )

    def forward(self, x):

        x = x.unsqueeze(1)

        x = self.transformer(x)

        return x.squeeze(1)
```

```python
class Projection(nn.Module):

    def __init__(self, input_dim, hidden_dim=256):

        super().__init__()

        self.layer = nn.Sequential(

            nn.Linear(input_dim, hidden_dim),

            nn.ReLU(),

            nn.Dropout(0.3)

        )

    def forward(self, x):

        return self.layer(x)
```

```python
class ContextTransformer(nn.Module):

    def __init__(self, dim=256):

        super().__init__()

        encoder = nn.TransformerEncoderLayer(

            d_model=dim,

            nhead=8,

            batch_first=True

        )

        self.transformer = nn.TransformerEncoder(

            encoder,

            num_layers=2
        )

    def forward(self, x):

        x = x.unsqueeze(1)

        x = self.transformer(x)

        return x.squeeze(1)
```

```python
class ModalityImportance(nn.Module):

    def __init__(self, hidden_dim=256):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(hidden_dim * 3, 256),

            nn.ReLU(),

            nn.Linear(256, 3),

            nn.Softmax(dim=1)

        )

    def forward(self, text, audio, video):

        x = torch.cat(
            [text, audio, video],
            dim=1
        )

        weights = self.network(x)

        return weights
```

```python
class CrossAttention(nn.Module):

    def __init__(self, dim=256):

        super().__init__()

        self.attention = nn.MultiheadAttention(
            embed_dim=dim,
            num_heads=8,
            batch_first=True
        )

    def forward(self, query, key, value):

        query = query.unsqueeze(1)
        key = key.unsqueeze(1)
        value = value.unsqueeze(1)

        output, _ = self.attention(
            query,
            key,
            value
        )

        return output.squeeze(1)
```

```python
class GatedFusion(nn.Module):

    def __init__(self, dim=256):

        super().__init__()

        self.gate = nn.Sequential(

            nn.Linear(dim * 2, dim),

            nn.Sigmoid()

        )

    def forward(self, x1, x2):

        gate = self.gate(
            torch.cat([x1, x2], dim=1)
        )

        fused = gate * x1 + (1 - gate) * x2

        return fused
```

```python
class DMIC2Model(nn.Module):

    def __init__(self, num_classes=3):

        super().__init__()

        # Projection Layers
        self.text_proj = Projection(256, 256)
        self.audio_proj = Projection(128, 256)
        self.video_proj = Projection(16, 256)

        # Context Transformers
        self.text_tf = ContextTransformer(256)
        self.audio_tf = ContextTransformer(256)
        self.video_tf = ContextTransformer(256)

        # Modality Importance Ranking
        self.mir = ModalityImportance(256)

        # Cross Attention
        self.cross = CrossAttention(256)

        # Gated Fusion
        self.gate = GatedFusion(256)

        # Final Classifier
        self.classifier = nn.Sequential(

            nn.Linear(256,128),

            nn.ReLU(),

            nn.Dropout(0.3),

            nn.Linear(128,num_classes)

        )

    def forward(self,text,audio,video):

        # Projection
        text=self.text_proj(text)
        audio=self.audio_proj(audio)
        video=self.video_proj(video)

        # Context Transformer
        text=self.text_tf(text)
        audio=self.audio_tf(audio)
        video=self.video_tf(video)

        # Dynamic Importance
        weights=self.mir(text,audio,video)

        wt=weights[:,0].unsqueeze(1)
        wa=weights[:,1].unsqueeze(1)
        wv=weights[:,2].unsqueeze(1)

        text=text*wt
        audio=audio*wa
        video=video*wv

        # Cascaded Cross Attention
        ta=self.cross(text,audio,audio)
        tav=self.cross(ta,video,video)

        # Gated Fusion
        fused=self.gate(ta,tav)

        output=self.classifier(fused)

        return output
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = DMIC2Model(
    num_classes=3
).to(device)

print(model)
```

```python
criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(

    model.parameters(),

    lr=5e-5

)

print("Ready for Training!")
```

```python
def train_one_epoch(model, loader, optimizer, criterion, device):

    model.train()

    running_loss = 0

    correct = 0

    total = 0

    for text, audio, video, labels in loader:

        text = text.to(device)
        audio = audio.to(device)
        video = video.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(text, audio, video)

        loss = criterion(outputs, labels)

        loss.backward()

        optimizer.step()

        running_loss += loss.item()

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)

        correct += (predicted == labels).sum().item()

    epoch_loss = running_loss / len(loader)

    epoch_acc = 100 * correct / total

    return epoch_loss, epoch_acc
```

```python
def evaluate(model, loader, criterion, device):

    model.eval()

    running_loss = 0

    correct = 0

    total = 0

    predictions = []

    targets = []

    with torch.no_grad():

        for text, audio, video, labels in loader:

            text = text.to(device)
            audio = audio.to(device)
            video = video.to(device)
            labels = labels.to(device)

            outputs = model(text, audio, video)

            loss = criterion(outputs, labels)

            running_loss += loss.item()

            _, predicted = torch.max(outputs, 1)

            total += labels.size(0)

            correct += (predicted == labels).sum().item()

            predictions.extend(predicted.cpu().numpy())

            targets.extend(labels.cpu().numpy())

    epoch_loss = running_loss / len(loader)

    epoch_acc = 100 * correct / total

    return epoch_loss, epoch_acc, predictions, targets
```

```python
print("Train Text :", train_text.shape)
print("Train Audio:", train_audio.shape)
print("Train Video:", train_video.shape)
print("Train Labels:", train_labels.shape)
```

```python
num_epochs = 20

best_acc = 0

for epoch in range(num_epochs):

    train_loss, train_acc = train_one_epoch(
        model,
        train_loader,
        optimizer,
        criterion,
        device
    )

    val_loss, val_acc, _, _ = evaluate(
        model,
        dev_loader,
        criterion,
        device
    )

    print(f"Epoch {epoch+1}/{num_epochs}")

    print(f"Train Loss : {train_loss:.4f}")

    print(f"Train Accuracy : {train_acc:.2f}%")

    print(f"Validation Loss : {val_loss:.4f}")

    print(f"Validation Accuracy : {val_acc:.2f}%")

    if val_acc > best_acc:

        best_acc = val_acc

        torch.save(
            model.state_dict(),
            "/content/drive/MyDrive/DMIC2/checkpoints/dmic2_best_model.pth"
        )

        print("✅ Best Model Saved!")

    print("-" * 50)
```
